## <a id="1"></a><font color="#2F855A">1. Estimadores</font>

<h3><font color="#004D40">Definición</font>  </3>

<div class="alert alert-block alert-info">
<b>Definición:</b> Un <b>estimador</b> es una regla, a menudo expresada como una fórmula, que indica cómo calcular el valor de una estimación con base en las mediciones contenidas en una muestra. 
    <div style="text-align: right; margin-top: 10px; font-size: 0.9em; color: #1A365D;">
        <i>(Wackerly et al., 2010, p. 391)</i>
    </div>
</div>

Un ejemplo de estimador puede ser la media muestral, dada por $$ \bar{Y} = \frac{1}{n}\sum_{i=1}^{n}Y_i $$, es un posible estimador de la media poblacional $\mu$.

<p>&#128161; Se pueden obtener muchos estimadores diferentes para el mismo parámetro poblacional, depende del razonamiento de la persona que esté definiendo la regla con la que estimará el parámetro requerido. Existen criterios para definir cuáles son los estimadores buenos. </p> 

<h3><font color="#004D40">Método de momentos</font>  </3>

El método de momentos "consiste en igualar los momentos muestrales con los correspondientes momentos poblacionales y resolver esta ecuación, o sistema de ecuaciones, para el parámetro o vector de parámetros, cuando ello sea posible" (Rincón, 2019, p.95).

<p>&#128161; Se igualan tantos momentos como parámetros haya que estimar. </p>

<h4><font color="#004D40">Ejemplo</font>  </4>

Sea $X_1, \dots, K_n$ una muestra aleatoria de una distribución normal con media $\mu$ y varianza $\sigma^2$. Sea $(\theta_1, \theta_2)=(\mu, \sigma)$. Estime los parámetros $\mu$ y $\sigma$ mediante el método de los momentos.

<b> Solución </b>

Recordemos que $$\sigma^2 = \mu_2'-  (\mu_1')^2$$ y $$\mu = \mu_1'$$.

Entonces las ecuaciones del método de los momentos se convierten en $$ M_1' = \mu_1' = \mu_1'(\mu, \sigma) = \mu $$, $$ M_2' = \mu_2' = \mu_2'(\mu, \sigma) = \sigma^2 + \mu^2 $$

y su solución es la siguiente:

El estimador de momentos de $\mu$ es $$ \hat{\mu}_{MM}=M_1'=\bar{X} $$, y el estimador de momentos es 
$$ \hat{\sigma}_{MM} = \sqrt{\frac{1}{n}\sum_{i=1}^{n}{X_i}^2-\bar{X}^2} $$

(Mood et al., 1974)

<h4><font color="#004D40">Ejemplo resuelto con python</font>  </4>

Para este ejemplo, se genera una muestra aleatoria con parámetros conocidos ($\mu_{real} = 10$, $\sigma_{real} = 2.5$) y luego se aplican los estimadores obtenidos ($\hat{\mu}_{MM}$ y $\hat{\sigma}_{MM}$) para recuperar esos valores a partir de los datos.

```python
import matplotlib.pyplot as plt
import numpy as np

# Configuración del experimento y generación de datos
np.random.seed(42)  # Semilla para que los resultados sean reproducibles

# Fijamos los parámetros teóricos reales de la población
mu_real = 10.0
sigma_real = 2.5
n = 1000  # Tamaño de la muestra (n)

# Generamos la muestra aleatoria X ~ N(mu_real, sigma_real^2)
X = np.random.normal(loc=mu_real, scale=sigma_real, size=n)

# Cálculo de los momentos muestrales (M1' y M2')
# Primer momento muestral: Promedio/Media muestral (X_bar)
M1_primo = np.mean(X)

# Segundo momento muestral: Promedio de los cuadrados (1/n * sum(X_i^2))
M2_primo = np.mean(X**2)

# Estimadores del Método de los Momentos (MM)
# Estimador para mu: \hat{\mu}_{MM} = M1' = X_bar
mu_hat_MM = M1_primo

# Estimador para sigma: \hat{\sigma}_{MM} = sqrt(M2' - (M1')^2)
varianza_hat_MM = M2_primo - (M1_primo**2)
sigma_hat_MM = np.sqrt(varianza_hat_MM)


#Impresión y comparación de resultados
print("   RESULTADOS DEL MÉTODO DE LOS MOMENTOS (MM)   ")
print(f"Tamaño de muestra (n): {n}\n")
print(f"Parámetro mu (Media):")
print(f"  • Real teórico : {mu_real:.4f}")
print(f"  • Estimado MM  : {mu_hat_MM:.4f}")
print(f"  • Error absoluto: {abs(mu_real - mu_hat_MM):.4f}\n")

print(f"Parámetro sigma (Desviación estándar):")
print(f"  • Real teórico : {sigma_real:.4f}")
print(f"  • Estimado MM  : {sigma_hat_MM:.4f}")
print(f"  • Error absoluto: {abs(sigma_real - sigma_hat_MM):.4f}")

# Visualización gráfica de la muestra y los parámetros
plt.figure(figsize=(8, 4.5), dpi=120)

# Histograma de los datos muestrales
plt.hist(
    X,
    bins=30,
    density=True,
    color="#0F4C81",
    alpha=0.65,
    edgecolor="#082947",
    label="Muestra observada",
)

# Curva de densidad teórica usando la estimación MM
x_axis = np.linspace(min(X), max(X), 200)
pdf_estimada = (1 / (sigma_hat_MM * np.sqrt(2 * np.pi))) * np.exp(
    -0.5 * ((x_axis - mu_hat_MM) / sigma_hat_MM) ** 2
)

plt.plot(
    x_axis,
    pdf_estimada,
    color="#E63946",
    linewidth=2.2,
    label=rf"Ajuste Normal MM $(\hat{{\mu}}={mu_hat_MM:.2f}, \hat{{\sigma}}={sigma_hat_MM:.2f})$",
)

plt.title(
    "Estimación de Parámetros por el Método de los Momentos",
    fontweight="bold",
    fontsize=11,
)
plt.xlabel("Valor de $X$", fontweight="bold")
plt.ylabel("Densidad de probabilidad", fontweight="bold")
plt.grid(True, linestyle="--", alpha=0.5)
plt.legend()
plt.tight_layout()
plt.show()
```

<h3><font color="#004D40">Estimación por máxima verosimilitud (EMV)</font>  </3>

<div class="alert alert-block alert-info">
<b>Definición:</b> La <b>función de verosimilitud</b> de un vector aleatorio $(X_1, \dots, X_n)$ cuya distribución depende de un parámetro $\theta$ se define como la función de densidad o de probabilidad conjunta $$ L(\theta) = f_{X_1,\dots, X_n}(X_1, \dots, X_n, \theta) $$
    <div style="text-align: right; margin-top: 10px; font-size: 0.9em; color: #1A365D;">
        <i>(Rincón, 2019, p. 108)</i>
    </div>
</div>

La estimación por máxima verosimilitud "consiste en encontrar el valor de $\theta$ que maximiza a la función $L(\theta)$. Al valor de $\theta$ en donde $L(\theta)$ alcanza su máximo se le llama la estimación de máxima verosmilitud o estimación de máximo verosímil" (Rincón, 2019, p. 109).

<h4><font color="#004D40">Ejemplo resuelto con python</font>  </4>

Lanzamos una moneda 100 veces, sabiendo que se obtienen 55 sole. Encontrar el EMV para la probabilidad $p$ de obtener sol en un solo lanzamiento.

Sea 
$X$: el número de soles que se obtienen al lanzar una moneda
se tiene que $X\sim \text{Bin}(100,p)$, entonces 

$$ f(x;p) = \binom{100}{55}p^{55}(1-p)^{100-55}$$

```python
from scipy.stats import binom
import numpy as np

# Vamos a utilizar la librería Sympy para matemática simbólica
import sympy

# importa la variable simbólica x
from sympy.abc import x
```

```python
# Vamos a definir a p somo una variable simbólica positiva: 0<p<1
p = sympy.symbols('p', positive= True)

# Definimos la función de verosimilitud
# se puede omitir la constante
f = p**55 * (1-p)**45

# Ahora derivamos f con respecto a p
df = sympy.diff(f,p)

# Hay que igualar la derivada a cero y despejar a p
# df=0 para encontrar p
phat = sympy.solve(df,p)[0] # [1] es la segunda solucion y [0] es la primera

print('El estimador de máxima verosimilitud es', phat)
```

La cual coincide con la proporción de soles.

<h4><font color="#004D40">Ejemplo resuelto con python</font>  </4>


Vamos a simular una muestra de tamaño 100 de ceros y unos.

```python
# n=1 : un ensayo de Bernoulli
# p : es la probabilidad de éxito
# size : genera 100 observaciones 
r = binom.rvs(n=1, p=0.7, size=100)
r
```

```python
# Vamos a calcular el EMV

# Definimos la función de verosimilitud
f= p**x * (1-p)**(1-x)

# Calculamos el producto de todos los elementos
# subs(x,i) sustituye variable simbólica x por un valor concreto de la muestra
J = np.prod([f.subs(x,i) for i in r])

# Calculamos el logaritmo logJ de manera expandida
# sympy.log(J) = logJ
logJ = sympy.expand_log(sympy.log(J))

# Calculamos la derivada e igualamos a cero
phat = sympy.solve(sympy.diff(logJ,p),p)[0]
print('El estimador de máxima verosimilitud es', phat)
```

```python
# Podemos verificar las soluciones con:
sol = sympy.solve(sympy.diff(logJ,p),p)
sol
```

## <a id="2"></a><font color="#2F855A">2. Propiedades de los estimadores</font>

<h3><font color="#004D40">Insesgadez</font>  </3>

<div class="alert alert-block alert-info">
<b>Definición:</b> Un estimador $\hat{\theta}$ es insesgado para el parámetro $\theta$ si se cumple la condición $$ E(\hat{\theta})=\theta $$
    <div style="text-align: right; margin-top: 10px; font-size: 0.9em; color: #1A365D;">
        <i>(Rincón, 2019, p. 127)</i>
    </div>
</div>

<p>&#128161; Esto quiere decir que el valor promedio es el valor a estimar. </p> 

<h3><font color="#004D40">Consistencia</font>  </3>

<div class="alert alert-block alert-info">
<b>Definición:</b> Sea $\hat{\theta}_n$ un estimador para $\theta$ basado en una muestra aleatoria de tamaño $n$. Se dice que $\hat{\theta}_n$ es <b>consistente</b> para $\theta$ si $\hat{\theta}_n \to \theta$ en probabilidad, cuando $n \to \infty$. Esto es, para cualquier $\epsilon > 0$, $$ \lim_{ n \to \infty} P(|\hat{\theta}_n-\theta|>\epsilon)=0 $$
    <div style="text-align: right; margin-top: 10px; font-size: 0.9em; color: #1A365D;">
        <i>(Rincón, 2019, p. 142)</i>
    </div>
</div>

Para poder determinar cuando un estimador es consistente, existe el siguiente criterio:

Sea $\hat{\theta}_n$ un estimador para $\theta$, basado en una muestra aleatoria de tamaño $n$. Si 

a) $ \lim_{n \to \infty} E(\hat{\theta}_n)=\theta$ y 

b) $ \lim_{n \to \infty} Var(\hat{\theta}_n)=0$,

entonces $\hat{\theta}_n$ es consistente.

<p>&#128161; Esto quiere decir que si un estimador es asintóticamente insesgado y su varianza tiende a cero, entonces es consistente. 

Hay que poner atención porque consistencia no implica insesgamiento.</p> 

<h3><font color="#004D40">Eficiencia</font>  </3>

<div class="alert alert-block alert-info">
<b>Definición:</b> Sean $\hat{\theta}_1$ y $\hat{\theta}_2$ dos estimadores insesgados para un parámetro $\theta$. Se dice que $\hat{\theta}_1$ es <b>relativamente más eficiente</b> que $\hat{\theta}_2$ si $$ Var(\hat{\theta}_1) \leq Var(\hat{\theta}_2)$$
    <div style="text-align: right; margin-top: 10px; font-size: 0.9em; color: #1A365D;">
        <i>(Rincón, 2019, p. 166)</i>
    </div>
</div>

<p>&#128161; Esto quiere decir que si se pueden comparar las varianzas de dos estimadores, se prefiere el que tenga la varianza más pequeña.

## <a id="3"></a><font color="#2F855A">3. Intervalos de confianza</font>

Un intervalo de confianza es un rango de valores que con cierta probabilidad (nivel de confianza) contiene al verdadero parámetro poblacional que queremos estimar.

Sea $X_1, \dots, X_n $ una muestra aleatoria con $X\sim\text{N}(\mu,\sigma^2)$, entonces

$$ ( num_1 < \mu < num_2) $$

En los intervalos de confianza, tenemos el error estándar (SE) el cual mide cuanto varia un estimador. 
- si conocemos $\sigma$, entonces: $SE = \sigma / \sqrt{n}$
- si NO conocemos $\sigma$, utilizamos $s$ y entonces: $SE = s/\sqrt{n}$

NOTA: a mayor tamaño de la muestra se tiene que el error estandar (SE) será menor y por lo tanto el intervalo de confianza será más preciso.

<div class="alert alert-block alert-info">
<b>Definición:</b> Si $X_1, \dots, X_n $ es una muestra aleatoria con densidad $f(x;\theta)$ y sean $T_1=t_1(X_1, \dots, X_n)$ y $T_2=t_2(X_1, \dots, X_n)$ dos estadísticos tal que $T_1 \le T_2$, entonces 
$$ \mathbb{P}(T_1 \le \theta \le T_2) = \gamma $$

Al intervalo $(T_1,T_2)$ se le llama intervalo del $\gamma\times 100\%$ de confianza. 
    <div style="text-align: right; margin-top: 10px; font-size: 0.9em; color: #1A365D;">
        <i>(Reyes, 2026)</i>
    </div>
</div>

<h3><font color="#004D40">Intervalo de confianza para una media</font>  </3>

Sea $X_1, \dots, X_n $ es una muestra aleatoria con $X\sim\text{N}(\mu,\sigma^2)$, entonces 
- $\bar{X_1} \sim \mathcal{N}(\mu, \sigma^2/n)$
- $Z = \frac{\bar{X_i} -\mu}{ \sigma/n} \sim \mathcal{N}(0,1) $

### $\sigma$ conocida:

Se tiene que el intervalo del $(1- \alpha ) \times 100\% $  de confianza para $\mu$ es:

$$ \bar{X} \pm z_{\alpha/2} \frac{\sigma}{\sqrt{n}} $$
donde
- $z_{\alpha/2} \frac{\sigma}{\sqrt{n}}$ es el margen de error
- $z_{\alpha/2}$ es el valor crítico 

**Ejemplo 1** Para tratar de estimar la media de consumo por cliente en un restaurante, se reunieron datos de una muestra de tamaño $49$ clientes durante tres semanas.

**a)** Supongamos que la desviación estándar poblacional es de $\$2.5$. ¿Cuál es el error estandar de la media?

**b)** Con un nivel de confianza del $95\%$, ¿cuál es el margen de error?

**c)** Si la media muestral es de $\$22.60$, ¿cuál es el intervalo de confianza del $95\%$ para la media poblacional?

```python
import numpy as np
import matplotlib.pyplot as plt
import scipy.stats as st

# Datos:
n = 49 # número de clientes
sigma = 2.5
media_muestral = 22.60
confianza = 0.95

# a) SE
error_est = sigma / np.sqrt(n)
print('El error estándar es:', error_est)

# b) 1-confianza = 0.95 -> 0.025 en cada cola -> (1-confianza)/2
z = st.norm.ppf(1-(1-confianza)/2)
margen_error = z*error_est
print('El margen de error es:', margen_error)

# c) Intervalo de confianza 
limite_inf = media_muestral - margen_error
limite_sup = media_muestral + margen_error
print(f'El intervalo del {confianza*100}% es: ({limite_inf:.2f}, {limite_sup:.2f})')
```

Por lo tanto, se tiene que el verdadero consumo poblacional en el restaurante se encuentra entre $ \$21.90 $ y $\$23.30 $

**Ejemplo 2** Supongamos que tenemos los siguientes datos de una muestra aleatoria de 10 observaciones:
$$ 45, 55, 67, 68, 45, 79, 98, 87, 84, 82 $$

Calcular un intervalo de confianza para la media.

```python
# Datos 
datos = [45, 55, 67, 68, 45, 79, 98, 87, 84, 82] # n<= 30
n = len(datos)
confianza = 0.99
gl = n-1

media_muestral = np.mean(datos)
error_est = st.sem(datos) # error estándar muestral 

# Vamos a utilizar la función 
# scipy.stats.t.interval(confidence, df, loc, scale)
# confidence = confianza
# df = degree freedom (gl)
# loc = media_muestral
# scale = error_est

intervalo = st.t.interval(confianza, gl, media_muestral, error_est)
print(f'El intervalo de confianza del {confianza*100}%: ({intervalo[0]:.2f}, {intervalo[1]:.2f})')
```

<h3><font color="#004D40">Intervalo de confianza para una proporción</font>  </3>

**Ejercicio 1** En una encuesta de una Universidad, 225 estudiantes son seleccionados al azar de los que toman cálculo y se les pregunta si cálculo es su materia favorita. 100 estudiantes responden que el cálculo es su materia favorita. Proporcione un intervalo de confianza del 95 %  para la proporción de todos los estudiantes que toman cálculo y que la consideran su materia favorita.

```python
import numpy as np
import scipy.stats as st

# Datos
n1 = 225  # total de estudiantes seleccionados
exitos1 = 100  # estudiantes que responden que cálculo es su materia favorita
confianza1 = 0.95
p_gorro1 = exitos1 / n1 # Proporción muestral

# Calculamos el Error estándar para una proporción (SE)
error_est1 = np.sqrt((p_gorro1 * (1 - p_gorro1)) / n1)
print("El error estándar es:", error_est1)

# Calculamos el Valor crítico z y el Margen de error
# 1-confianza = 0.95 -> 0.025 en cada cola -> (1-confianza)/2
z1 = st.norm.ppf(1 - (1 - confianza1) / 2)
margen_error1 = z1 * error_est1
print("El margen de error es:", margen_error1)

# Calculamos el Intervalo de confianza
limite_inf1 = p_gorro1 - margen_error1
limite_sup1 = p_gorro1 + margen_error1
print(f"El intervalo del {confianza1*100}%  de confianza es: ({limite_inf1:.4f}, {limite_sup1:.4f}) ")
```

<h3><font color="#004D40">Interpretación del intervalo</font>  </3>

Al analizar las respuestas de los 225 estudiantes, pudimos estimar con un 95% de seguridad que la verdadera proporción de alumnos que aman el cálculo está entre el 37.95% y el 50.94%.

## <a id="4"></a><font color="#2F855A">4. Conclusiones</font>

Al hacer estadística inferencial, pasamos de simplemente describir lo que vemos en un grupo pequeño a intentar "adivinar" con fundamentos científicos qué está pasando en toda una población. Debido a que no siempre es posible encuestar a la totalidad de una población estudiada, las herramientas de la estadística inferencial estudiadas en este Notebook son vitales para manejar la incertidumbre y el azar que rodean la realidad humana.

Los estimadores funcionan como una "regla" o un puente que nos permite usar los datos de una muestra para calcular una característica que no conocemos del grupo completo, básicamente, son los encargados de convertir la información limitada que tenemos en una afirmación sobre la población objetivo.

Debido a que hay tantos estimadores como maneras de pensar de quienes resuelvan un problema, es necesario saber si un estimador es "bueno", para ello evaluamos sus propiedades:

- <b>Insesgadez</b>, que significa que, si repitiéramos el proceso muchas veces, el promedio de nuestras estimaciones daría exactamente el valor real que buscamos; es decir, que no tenemos un error sistemático hacia arriba o hacia abajo.
- <b>Eficiencia</b>, prefiriendo siempre aquel método que varíe lo menos posible, para que nuestros resultados no den saltos tan grandes de una muestra a otra.
- <b>Consistencia</b>, que nos da la tranquilidad de saber que, conforme logramos obtener más datos y nuestra muestra crece, nuestra estimación se acercará cada vez más a la verdad.

Finalmente, los intervalos de confianza son la forma más honesta de presentar los resultados. En lugar de dar un solo número con certeza, ofrecemos un rango de valores razonables donde creemos que se encuentra el parámetro real.

Su interpretación es muy interesante: un nivel de confianza del 95% no significa que hay un 95% de probabilidad de que el valor esté ahí dentro hoy, sino que el procedimiento es tan confiable que, si hiciéramos el estudio 100 veces, en 95 de ellas el rango obtenido atraparía el valor verdadero de la población.
