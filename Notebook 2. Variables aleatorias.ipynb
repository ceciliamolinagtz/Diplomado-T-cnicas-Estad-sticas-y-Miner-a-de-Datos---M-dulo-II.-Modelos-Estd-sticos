#### ¿QUÉ ES UNA VARIABLE ALEATORIA?

Al hacer un experimento, el resultado directo puede ser algo no númerico, como 'sol' o 'águila' en el caso de lanzar una moneda. Una **variable aleatoria** es una regla o función que asigna un número real a cada resultado posible de ese experimento.

Matemáticamente, se define como una transformación $X$ del espacio muestral $\Omega$ a $\mathbb{R}$, es decir, $X: \Omega \to \mathbb{R}$.

Principalmente pueden dividirse en:
- **Discretas:** toman un número finito o numerable de valores.
- **Continuas:** Pueden tomar cualquier valor en un intervalo.

## <a id="1"></a><font color="#2F855A">1. Variables aleatorias discretas</font>

<div class="alert alert-block alert-info">
    <b>Definición:</b> Se dice que una variable aleatoria $Y$ es <b>discreta</b> si puede tomar sólo un número finito o contablemente infinito de valores distintos.
    <div style="text-align: right; margin-top: 10px; font-size: 0.9em; color: #1A365D;">
        <i>(Wackerly, Mendenhall, Scheaff, xx, p. 87)</i>
    </div>
</div>

Como ciertos tipos de variables aleatorias se presentan frecuentemente en la práctica, es útil tener a la mano la probabilidad para cada valor de una variable aleatoria. Este conjunto de probabilidades se conoce como <b> distribución de probabilidad </b> de la variable aleatoria discreta.

<div class="alert alert-block alert-info">
    <b>Definición:</b> La probabilidad de que $Y$ tome el valor $y$, $P(Y=y)$, se define como la suma de las probabilidades de todos los puntos muestrales en S a los que se asigna el valor $y$. A veces denotaremos $P(Y=y)$ por $p(y)$.
    <div style="text-align: right; margin-top: 10px; font-size: 0.9em; color: #1A365D;">
        <i>(Wackerly, Mendenhall, Scheaff, xx, p. 88)</i>
    </div>
</div>

### <a id="2"></a><font color="#004D40">1.1 Distribución Uniforme Discreta</font>

Cuando tenemos $n$ resultados diferentes y todos tienen la misma probabilidad de ocurrir se da la **distribución uniforme**.

<div class="alert alert-block alert-info">
<b>Definición:</b> Decimos que una variable aleatoria $X$ tiene una distribución uniforme discreta sobre el conjunto de $n$ números ${x_1, \dots,x_n}$ si la probabilidad de que $X$ tome cualquiera de estos valores es constante $1/n$. 
    <div style="text-align: right; margin-top: 10px; font-size: 0.9em; color: #1A365D;">
        <i>(Rincón, 2014, p. 212)</i>
    </div>
</div>

Esto quiere decir que podemos expresar la **función de probabilidad** como 
$$f_{X}(k)=\mathbb{P}(X=k)=\begin{cases}
\frac{1}{n} &\quad \text{si } k\in\{x_{1},\dots,x_{n}\} \\
0 &\quad \text{en otro caso}
\end{cases}$$

#### ESPERANZA Y VARIANZA DE LA DISTRIBUCIÓN UNIFORME DISCRETA

Si se tiene una distribución uniforme discreta, en este caso, se tiene que la esperanza y varianza están dadas por:
	$$\mathbb{E}[X] = \sum_{k}kf_{X}(k)=\frac{x_n + x_1}{2}.$$
	y
	$$\text{Var}(X) = \mathbb{E}[(X-\mathbb{E}[X])^{2}] =\sum_{k}(k-\mathbb{E}[X])^{2}f_{X}(k)=\frac{(x_n - x_1+1)^2-1}{12}$$

#### VISUALIZACIÓN EN PYTHON

```python
# Distribución uniforme discreta
min = 0
max = 20
x = np.arange(min, max)

# Vamos a crear la función de densidad (pmf (probability mass function))
pmf_unif = np.full_like(x, 1/(max-min), dtype=float) # crear un arreglo del mismo tamaño

#Graficamos la función de densidad
plt.figure(figsize=(6,4))
plt.bar(x, pmf_unif, color='#2F855A80', edgecolor='#004D40') # grafica de barras
plt.xlabel("Valores")
plt.ylabel("Probabilidad")
plt.title("Función de densidad de una variable aleatoria uniforme discreta")
plt.grid(True)
plt.show()
```

```python
import numpy as np
import plotly.graph_objects as go

# Fijamos una semilla para que los resultados sean reproducibles
np.random.seed(123)

# Parámetros del rango (valores enteros posibles del dado/urna, de 'a' a 'b')
a = 1  # Límite inferior (inclusive)
b = 6  # Límite superior (inclusive)

# Número de simulaciones
n_simulaciones = 10000

# Generamos las simulaciones usando NumPy
# Nota: randint va de 'a' (inclusivo) a 'b + 1' (exclusivo)
datos_uniforme_disc = np.random.randint(a, b + 1, n_simulaciones)

# Rango de valores posibles enteros
x_vals = np.arange(a, b + 1)
# En una uniforme discreta todos los valores tienen exactamente la misma probabilidad
total_valores = b - a + 1
y_vals = np.full(total_valores, 1 / total_valores)

fig = go.Figure()

# Agregamos el histograma de la simulación empírica
fig.add_trace(go.Histogram(
    x = datos_uniforme_disc,
    histnorm = 'probability',
    marker_color = 'skyblue',
    name = 'Simulación',
    # Centramos las barras perfectamente sobre cada número entero
    xbins = dict(start=a - 0.5, end=b + 0.5, size=1)
))

# Agregamos la distribución teórica discreta (línea horizontal con marcadores)
fig.add_trace(go.Scatter(
    x = x_vals,
    y = y_vals,
    mode = 'lines+markers',
    name = 'PMF Teórica',
    line = dict(color='darkblue', width=3, dash='dash'), # Línea punteada para denotar constancia
    marker = dict(size=6)
))

# Configuración y estética de la gráfica interactiva
fig.update_layout(
    title = f"Distribución Uniforme Discreta Simulada vs Teórica (Valores del {a} al {b})",
    xaxis_title = 'Resultado (x)',
    yaxis_title = 'Probabilidad',
    bargap = 0.1,
    xaxis = dict(range=[a - 0.5, b + 0.5], tickmode = 'linear', tick0 = a, dtick = 1)
)

fig.show()
```

### <a id="3"></a><font color="#004D40">1.2 Distribución de Bernoulli</font>

Modela un experimento con solo dos resultados posibles: 'éxito' (1) y 'fracaso' (0).

<div class="alert alert-block alert-info">
    <b>Definición:</b> Un <b> ensayo de Bernoulli</b> se define como aquel experimento aleatorio con únicamente dos posibles resultados, llamados genéricamente: éxito y fracaso. Supondremos que las probabilidades de estos resultados son $p$ y $1-p$, respectivamente. Si se define la variable aleatoria $X$ como aquella función que lleva el resultado éxito al número 1 y el resultado fracaso al número 0, entonces decimos que $X$ tiene una <b>distribución Bernoulli</b> con parámetro $p \in (0,1)$. La función de probabilidad se puede escribir de la siguiente forma.
    $$f(x)= \begin{cases}
    1-p & \text{si  } x=0 \\
    p & \text{si  } x=1 \\
    0 & \text{en otro caso} 
    \end{cases}$$
    <div style="text-align: right; margin-top: 10px; font-size: 0.9em; color: #1A365D;">
        <i>(Rincón, 2014, p. 217)</i>
    </div>
</div>

```python
import numpy as np
import plotly.graph_objects as go

# Fijamos una semilla para que los resultados sean reproducibles
np.random.seed(123)

# Probabilidad de éxito (p)
p_exito = 0.6

# Número de simulaciones
n_simulaciones = 10000

# Generamos las simulaciones (Bernoulli es una Binomial con n=1)
datos_bernoulli = np.random.binomial(1, p_exito, n_simulaciones)

# Definimos los dos únicos valores posibles: 0 y 1
x_vals = np.array([0, 1])
y_vals = np.array([1 - p_exito, p_exito]) # PMF Teórica

fig = go.Figure()

# Agregamos el histograma de la simulación empírica
fig.add_trace(go.Histogram(
    x = datos_bernoulli,
    histnorm = 'probability',
    marker_color = 'skyblue',
    name = 'Simulación',
    # Centramos las dos barras perfectamente en 0 y 1
    xbins = dict(start=-0.5, end=1.5, size=1)
))

# Agregamos la distribución teórica discreta (línea con marcadores sobre el 0 y el 1)
fig.add_trace(go.Scatter(
    x = x_vals,
    y = y_vals,
    mode = 'markers',
    name = 'PMF Teórica',
    marker = dict(color='darkblue', size=12, symbol='diamond')
))

# Configuración y estética de la gráfica interactiva
fig.update_layout(
    title = f"Distribución de Bernoulli Simulada vs Teórica (p = {p_exito})",
    xaxis_title = 'Resultado (0 = Fracaso, 1 = Éxito)',
    yaxis_title = 'Probabilidad',
    bargap = 0.6, # Hacemos las barras más esbeltas al ser solo dos categorías
    xaxis = dict(tickmode = 'array', tickvals = [0, 1], range = [-0.6, 1.6])
)

fig.show()
```

### <a id="4"></a><font color="#004D40">1.3 Distribución Binomial</font>

Para experimentos que consisten en la observación de experimentos idénticos e independientes, cada uno de los cuáles puede resultar en una de dos salidas.

Por ejemplo:
- Artículos saliendo de línea de producción: pueden ser '**defectuoso**' o '**no defectuoso**'.
- Disparos en tiro al blanco: pueden ser '**acierto**' o '**no acierto**'.
- Encuesta para ver quienes están a favor de un candidato: '**a favor**' o '**en contra**'. 

#### PROPIEDADES DE UN EXPERIMENTO BINOMIAL
1. Consiste en un número fijo $n$ de pruebas idénticas.
2. Cada prueba resulta en uno de dos resultados: éxito S, o fracaso F.
3. La probabilidad de éxito en una sola prueba es igual a algún valor $p$ y es el mismo de una prueba a la otra. La probabilidad de fracaso es igual a $q=(1-p)$.
4. Las pruebas son independientes.
5. La variable aleatoria de interés es $Y$, el número de éxitos observados durante las $n$ pruebas.

<div class="alert alert-block alert-info">
    <b>Definición:</b> Se dice que una variable aleatoria $Y$ tiene una <b>distribución binomial</b> basada en $n$ pruebas con probabilidad $p$ de éxito si y sólo si 
    \begin{equation}
    p(y) = \binom{n}{y}p^yq^{n-y}, \qquad y=0,1,2, \dots, n \quad \text{ y } \quad  0\leq p \leq 1.
    \end{equation}
    <div style="text-align: right; margin-top: 10px; font-size: 0.9em; color: #1A365D;">
        <i>(Wackerly, Mendenhall, Scheaff, xx, p. 103)</i>
    </div>
</div>

#### ESPERANZA y VARIANZA DE UNA VARIABLE ALEATORIA BINOMIAL

Sea $Y$ una variable aleatoria binomial basada en $n$ pruebas y probabilidad $p$ de éxito. Entonces 
$$ \mu = E(Y)= np$$ y $$ \sigma^2 = V(Y) = npq$$

#### VISUALIZACIÓN EN PYTHON

```python
# Grafica de la distribución Binomial
# Parámetros
n = 50
p = 0.5

# Generamos numeros aleatorios con distribución binomial
binom_numeros = sps.binom.rvs(n,p,size = 10000)

# Creamos un histograma
plt.figure(figsize=(8,4))
plt.hist(
    binom_numeros,
    density = True, # normaliza para que sea función de densidad
    bins = len(np.unique(binom_numeros)), # numero de barras en el histograma
    color = "#2F855A80",
    edgecolor = '#004D40'         
)
plt.xlabel('Valores')
plt.ylabel('Probabilidad')
plt.title('Densidad de una variable aleatoria binomial')
plt.show()
```

```python
import numpy as np
import plotly.graph_objects as go
from scipy.stats import binom

# Fijamos una semilla para que los resultados sean reproducibles
np.random.seed(123)

# Parámetros de la distribución binomial
n_ensayos = 20    # Número total de ensayos independientes (n)
p_exito = 0.4      # Probabilidad de éxito en cada ensayo (p)

# Número de simulaciones
n_simulaciones = 10000

# Generamos las simulaciones usando NumPy
datos_binomiales = np.random.binomial(n_ensayos, p_exito, n_simulaciones)

# Definimos el rango de todos los valores posibles (de 0 a n ensayos)
x_vals = np.arange(0, n_ensayos + 1)
y_vals = binom.pmf(x_vals, n_ensayos, p_exito)

fig = go.Figure()

# Agregamos el histograma de la simulación empírica
fig.add_trace(go.Histogram(
    x = datos_binomiales,
    histnorm = 'probability',
    marker_color = 'skyblue',
    name = 'Simulación',
    # Alineamos las barras perfectamente con los números enteros
    xbins = dict(start=-0.5, end=n_ensayos + 0.5, size=1)
))

# Agregamos la distribución teórica discreta (línea con marcadores)
fig.add_trace(go.Scatter(
    x = x_vals,
    y = y_vals,
    mode = 'lines+markers',
    name = 'PMF Teórica',
    line = dict(color='darkblue', width=3),
    marker = dict(size=6)
))

# Configuración y estética de la gráfica interactiva
fig.update_layout(
    title = f"Distribución Binomial Simulada vs Teórica (n = {n_ensayos}, p = {p_exito})",
    xaxis_title = 'Número de éxitos (x)',
    yaxis_title = 'Probabilidad',
    bargap = 0.1,
    xaxis = dict(range=[-0.5, n_ensayos + 0.5], tickmode = 'linear', tick0 = 0, dtick = 2)
)

fig.show()
```

### <a id="5"></a><font color="#004D40">1.4 Distribución Geométrica</font>

<div class="alert alert-block alert-info">
    <b>Definición:</b> Decimos que $X$ tiene una distribución geométrica con parámetro $p$ y escribimos $X \sim geo(p)$ cuando su función de probabilidad es
    $$f(x)= \begin{cases}
    p(1-p)^x & \text{si  } x=0, 1, \dots \\
    0 & \text{en otro caso} 
    \end{cases}$$
    <div style="text-align: right; margin-top: 10px; font-size: 0.9em; color: #1A365D;">
        <i>(Rincón, 2014, p. 230)</i>
    </div>
</div>

```python
import numpy as np
import plotly.graph_objects as go
from scipy.stats import geom

# Fijamos una semilla para que los resultados sean reproducibles
np.random.seed(123)

# Probabilidad de éxito en cada ensayo independiente (p)
p_exito = 0.3

# Número de simulaciones
n_simulaciones = 10000

# Generamos las simulaciones usando NumPy (cuenta el número de ensayos totales)
datos_geometricos = np.random.geometric(p_exito, n_simulaciones)

# Definimos el rango de valores enteros para la PMF teórica
# Usamos el percentil 99.9 de los datos para recortar la cola larga de la gráfica de forma limpia
max_val = int(np.percentile(datos_geometricos, 99.9))
x_vals = np.arange(1, max_val + 1)
y_vals = geom.pmf(x_vals, p_exito)

fig = go.Figure()

# Agregamos el histograma de la simulación empírica
fig.add_trace(go.Histogram(
    x = datos_geometricos,
    histnorm = 'probability',
    marker_color = 'skyblue',
    name = 'Simulación',
    # Alineamos las barras perfectamente con los números enteros a partir de 1
    xbins = dict(start=0.5, end=max_val + 0.5, size=1)
))

# Agregamos la distribución teórica discreta (línea con marcadores)
fig.add_trace(go.Scatter(
    x = x_vals,
    y = y_vals,
    mode = 'lines+markers',
    name = 'PMF Teórica',
    line = dict(color='darkblue', width=3),
    marker = dict(size=6)
))

# Configuración y estética de la gráfica interactiva
fig.update_layout(
    title = f"Distribución Geométrica (Ensayos hasta el Primer Éxito, p = {p_exito})",
    xaxis_title = 'Número de ensayos (x)',
    yaxis_title = 'Probabilidad',
    bargap = 0.1,
    xaxis = dict(range=[0.5, max_val + 0.5], tickmode = 'linear', tick0 = 1, dtick = 1)
)

fig.show()
```

### <a id="6"></a><font color="#004D40">1.5 Distribución de Poisson</font>

### <a id="7"></a><font color="#004D40">1.6 Distribución Binomial negativa</font>

<div class="alert alert-block alert-info">
    <b>Definición:</b> Decimos que $X$ tiene una distribución binomial negativa con parámtros $r$ y $p$, y escribimos $X \sim bin neg(r,p)$. Es claro que la variable $X$ puede tomar los valores $0,1,2,\dots$ con las probabilidades dadas por la función de probabilidad
    $$f(x)= \begin{cases}
    \binom{r+x-1}{x}p^r(1-p)^x & \text{si} x=0,1,\dots \\
    0 & \text{en otro caso} 
    \end{cases}$$
    <div style="text-align: right; margin-top: 10px; font-size: 0.9em; color: #1A365D;">
        <i>(Rincón, 2014, p. 237)</i>
    </div>
</div>

```python
import numpy as np
import plotly.graph_objects as go
from scipy.stats import nbinom

# Fijamos una semilla para que los resultados sean reproducibles
np.random.seed(123)

# Parámetros de la distribución binomial negativa
n_exitos_requeridos = 5  # Número de éxitos que queremos alcanzar (parámetro r o n)
p_exito = 0.3            # Probabilidad de éxito en cada ensayo independiente (p)

# Número de simulaciones
n_simulaciones = 10000

# Generamos las simulaciones usando NumPy (cuenta el número de fracasos obtenidos)
datos_nbinom = np.random.negative_binomial(n_exitos_requeridos, p_exito, n_simulaciones)

# Definimos el rango de valores (fracasos) para la PMF teórica
# Usamos el percentil 99.9 de los datos para que la gráfica no se extienda infinitamente a la derecha
max_val = int(np.percentile(datos_nbinom, 99.9))
x_vals = np.arange(0, max_val + 1)
y_vals = nbinom.pmf(x_vals, n_exitos_requeridos, p_exito)

fig = go.Figure()

# Agregamos el histograma de la simulación empírica
fig.add_trace(go.Histogram(
    x = datos_nbinom,
    histnorm = 'probability',
    marker_color = 'skyblue',
    name = 'Simulación',
    # Alineamos las barras con los números enteros de fracasos
    xbins = dict(start=-0.5, end=max_val + 0.5, size=1)
))

# Agregamos la distribución teórica discreta (línea con marcadores)
fig.add_trace(go.Scatter(
    x = x_vals,
    y = y_vals,
    mode = 'lines+markers',
    name = 'PMF Teórica',
    line = dict(color='darkblue', width=3),
    marker = dict(size=6)
))

# Configuración y estética de la gráfica interactiva
fig.update_layout(
    title = f"Distribución Binomial Negativa (Número de Fracasos hasta {n_exitos_requeridos} Éxitos, p={p_exito})",
    xaxis_title = 'Número de fracasos (x)',
    yaxis_title = 'Probabilidad',
    bargap = 0.1,
    xaxis = dict(range=[-0.5, max_val + 0.5]) # Limitamos el eje X para que se vea limpio
)

fig.show()
```

### <a id="8"></a><font color="#004D40">1.7 Distribución Hipergeométrica</font>

```python
import numpy as np
import plotly.graph_objects as go
from scipy.stats import hypergeom

# Fijamos una semilla para que los resultados sean reproducibles
np.random.seed(123)

# Parámetros de la distribución hipergeométrica:
# Ejemplo: Una urna con 50 canicas en total (M), 20 son rojas (n) y extraemos 10 sin reemplazo (N)
M_total = 50       # Número total de elementos en la población
n_exitos = 20      # Número total de elementos con la característica deseada (éxitos en la población)
N_muestra = 10     # Tamaño de la muestra extraída

# Número de simulaciones
n_simulaciones = 10000

# Generamos las simulaciones usando NumPy
# Nota: NumPy pide (ngood, nbad, nsample) -> éxitos, fracasos, tamaño de muestra
datos_hipergeom = np.random.hypergeometric(n_exitos, M_total - n_exitos, N_muestra, n_simulaciones)

# Definimos el rango de valores posibles para la PMF teórica (valores enteros)
x_vals = np.arange(0, N_muestra + 1)
y_vals = hypergeom.pmf(x_vals, M_total, n_exitos, N_muestra)

fig = go.Figure()

# Agregamos el histograma de la simulación empírica
fig.add_trace(go.Histogram(
    x = datos_hipergeom,
    histnorm = 'probability',
    marker_color = 'skyblue',
    name = 'Simulación',
    # Forzamos a que las barras se alineen perfectamente con los números enteros
    xbins = dict(start=-0.5, end=N_muestra + 0.5, size=1)
))

# Agregamos la distribución teórica discreta (línea con marcadores sobre los enteros)
fig.add_trace(go.Scatter(
    x = x_vals,
    y = y_vals,
    mode = 'lines+markers',
    name = 'PMF Teórica',
    line = dict(color='darkblue', width=3),
    marker = dict(size=8)
))

# Configuración y estética de la gráfica interactiva
fig.update_layout(
    title = f"Distribución Hipergeométrica Simulada vs Teórica (M={M_total}, n={n_exitos}, N={N_muestra})",
    xaxis_title = 'Número de éxitos (x)',
    yaxis_title = 'Probabilidad',
    bargap = 0.1,
    xaxis = dict(tickmode = 'linear', tick0 = 0, dtick = 1) # Asegura marcas en enteros
)

fig.show()
```

### <a id="10"></a><font color="#004D40">2.1 Distribución Uniforme Continua</font>

```python
# Fijamos una semilla para que los resultados sean reproducibles
np.random.seed(123)

#  Definimos los parámetros del intervalo [a, b] para la Uniforme Continua
a_param = 5
b_param = 15

# Número de simulaciones (usamos 10,000 para que se vea genial)
n = 10000

# Generamos n números uniformes estándar en el intervalo (0, 1)
U = np.random.random(n)

#  Aplicamos la ecuación de la transformación inversa derivada en clase:
# F(x) = (x - a) / (b - a) = U  =>  X = a + (b - a) * U
uniformes_continuas = a_param + (b_param - a_param) * U

#  Creamos la función de densidad teórica f(x) = 1 / (b - a)
x_vals = np.linspace(a_param - 2, b_param + 2, 500)
y_vals = np.where((x_vals >= a_param) & (x_vals <= b_param), 1 / (b_param - a_param), 0)

fig = go.Figure()

# Agregamos el histograma detallado con bastantes barras (nbinsx = 40)
fig.add_trace(go.Histogram(
    x = uniformes_continuas,
    nbinsx = 40,
    histnorm = 'probability density',
    marker_color = 'skyblue',
    name = 'Simulación'
))

# Agregamos la línea de la densidad teórica en color azul oscuro
fig.add_trace(go.Scatter(
    x = x_vals,
    y = y_vals,
    mode = 'lines',
    name = 'Densidad teórica',
    line = dict(color='darkblue', width=3)
))

# Configuración y estética de la gráfica interactiva
fig.update_layout(
    title = "Distribución Uniforme Continua Simulada vs Teórica Unif(5, 15)",
    xaxis_title = 'x',
    yaxis_title = 'Densidad',
    bargap = 0.05
)

fig.show()
```

### <a id="11"></a><font color="#004D40">2.2 Distribución Normal</font>

```python
import numpy as np
import plotly.graph_objects as go
from scipy.stats import norm

# Fijamos una semilla para que los resultados sean reproducibles
np.random.seed(123)

# Definimos los parámetros de la Distribución Normal (Media y Desviación Estándar)
mu = 0
sigma = 1

# Número de simulaciones
n = 10000

# Generamos n números con distribución normal
datos_normales = np.random.normal(mu, sigma, n)

# Creamos la función de densidad teórica usando scipy.stats.norm
x_vals = np.linspace(mu - 4*sigma, mu + 4*sigma, 500)
y_vals = norm.pdf(x_vals, mu, sigma)

fig = go.Figure()

# Agregamos el histograma detallado con bastantes barras (nbinsx = 40)
fig.add_trace(go.Histogram(
    x = datos_normales,
    nbinsx = 40,
    histnorm = 'probability density',
    marker_color = 'skyblue',
    name = 'Simulación'
))

# Agregamos la línea de la densidad teórica en color azul oscuro
fig.add_trace(go.Scatter(
    x = x_vals,
    y = y_vals,
    mode = 'lines',
    name = 'Densidad teórica',
    line = dict(color='darkblue', width=3)
))

# Configuración y estética de la gráfica interactiva
fig.update_layout(
    title = f"Distribución Normal (Gaussiana) Simulada vs Teórica N({mu}, {sigma}^2)",
    xaxis_title = 'x',
    yaxis_title = 'Densidad',
    bargap = 0.05
)

fig.show()
```

### <a id="13"></a><font color="#004D40">2.4 Distribución Gamma</font>

```python
import numpy as np
import plotly.graph_objects as go
from scipy.stats import gamma

# Fijamos una semilla para que los resultados sean reproducibles
np.random.seed(123)

# Definimos los parámetros de la distribución Gamma
shape_k = 2       # Parámetro de forma (k o alpha)
scale_theta = 2   # Parámetro de escala (theta o 1/beta)

# Número de simulaciones
n = 10000

# Generamos n números con distribución Gamma
datos_gamma = np.random.gamma(shape_k, scale_theta, n)

# Creamos la función de densidad teórica usando scipy.stats.gamma
x_vals = np.linspace(0, np.max(datos_gamma), 500)
y_vals = gamma.pdf(x_vals, a=shape_k, scale=scale_theta)

fig = go.Figure()

# Agregamos el histograma detallado con bastantes barras (nbinsx = 40)
fig.add_trace(go.Histogram(
    x = datos_gamma,
    nbinsx = 40,
    histnorm = 'probability density',
    marker_color = 'skyblue',
    name = 'Simulación'
))

# Agregamos la línea de la densidad teórica en color azul oscuro
fig.add_trace(go.Scatter(
    x = x_vals,
    y = y_vals,
    mode = 'lines',
    name = 'Densidad teórica',
    line = dict(color='darkblue', width=3)
))

# Configuración y estética de la gráfica interactiva
fig.update_layout(
    title = f"Distribución Gamma Simulada vs Teórica Gamma(k={shape_k}, θ={scale_theta})",
    xaxis_title = 'x',
    yaxis_title = 'Densidad',
    bargap = 0.05
)

fig.show()
```
