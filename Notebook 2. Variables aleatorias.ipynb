### <font color='2F855A'>¿Qué es una variable aleatoria?<font>

Al hacer un experimento, el resultado directo puede ser algo no númerico, como 'sol' o 'águila' en el caso de lanzar una moneda. Una **variable aleatoria** es una regla o función que asigna un número real a cada resultado posible de ese experimento.

Matemáticamente, se define como una transformación $X$ del espacio muestral $\Omega$ a $\mathbb{R}$, es decir, $X: \Omega \to \mathbb{R}$.

Principalmente pueden dividirse en:
- **Discretas:** toman un número finito o numerable de valores.
- **Continuas:** Pueden tomar cualquier valor en un intervalo.

## <a id="1"></a><font color="#2F855A">1. Variables aleatorias discretas</font>

<div class="alert alert-block alert-info">
    <b>Definición:</b> Se dice que una variable aleatoria $Y$ es <b>discreta</b> si puede tomar sólo un número finito o contablemente infinito de valores distintos.
    <div style="text-align: right; margin-top: 10px; font-size: 0.9em; color: #1A365D;">
        <i>(Wackerly, Mendenhall, Scheaff, xx, p. 87)</i>
    </div>
</div>

Como ciertos tipos de variables aleatorias se presentan frecuentemente en la práctica, es útil tener a la mano la probabilidad para cada valor de una variable aleatoria. Este conjunto de probabilidades se conoce como <b> distribución de probabilidad </b> de la variable aleatoria discreta.

<div class="alert alert-block alert-info">
    <b>Definición:</b> La probabilidad de que $Y$ tome el valor $y$, $P(Y=y)$, se define como la suma de las probabilidades de todos los puntos muestrales en S a los que se asigna el valor $y$. A veces denotaremos $P(Y=y)$ por $p(y)$.
    <div style="text-align: right; margin-top: 10px; font-size: 0.9em; color: #1A365D;">
        <i>(Wackerly, Mendenhall, Scheaff, xx, p. 88)</i>
    </div>
</div>

### <a id="2"></a><font color="#004D40">1.1 Distribución Uniforme Discreta</font>

Cuando tenemos $n$ resultados diferentes y todos tienen la misma probabilidad de ocurrir se da la **distribución uniforme**.

<div class="alert alert-block alert-info">
<b>Definición:</b> Decimos que una variable aleatoria $X$ tiene una distribución uniforme discreta sobre el conjunto de $n$ números ${x_1, \dots,x_n}$ si la probabilidad de que $X$ tome cualquiera de estos valores es constante $1/n$. 
    <div style="text-align: right; margin-top: 10px; font-size: 0.9em; color: #1A365D;">
        <i>(Rincón, 2014, p. 212)</i>
    </div>
</div>

Esto quiere decir que podemos expresar la **función de probabilidad** como 
$$f_{X}(k)=\mathbb{P}(X=k)=\begin{cases}
\frac{1}{n} &\quad \text{si } k\in\{x_{1},\dots,x_{n}\} \\
0 &\quad \text{en otro caso}
\end{cases}$$

#### ESPERANZA Y VARIANZA DE LA DISTRIBUCIÓN UNIFORME DISCRETA

Si se tiene una distribución uniforme discreta, en este caso, se tiene que la esperanza y varianza están dadas por:
	$$\mathbb{E}[X] = \sum_{k}kf_{X}(k)=\frac{x_n + x_1}{2}.$$
	y
	$$\text{Var}(X) = \mathbb{E}[(X-\mathbb{E}[X])^{2}] =\sum_{k}(k-\mathbb{E}[X])^{2}f_{X}(k)=\frac{(x_n - x_1+1)^2-1}{12}$$

<h3>&#128202; <font color="#004D40">Visualización en Python</font>  </3>

```python
import matplotlib.pyplot as plt
import numpy as np
import seaborn as sns

# Configuración del estilo general de la gráfica para GitHub/Academic Notebooks
sns.set_theme(style="whitegrid", palette="muted")
plt.rcParams.update(
    {
        "font.family": "sans-serif",
        "font.size": 10,
        "axes.labelsize": 11,
        "axes.titlesize": 13,
        "xtick.labelsize": 9,
        "ytick.labelsize": 9,
        "figure.dpi": 150,  # Alta resolución para renderizado fluido en GitHub
    }
)

# 1. Definición de parámetros del espacio muestral
min_val = 0
max_val = 20
x = np.arange(min_val, max_val)
n_obs = len(x)

# 2. Función de Masa de Probabilidad (PMF)
pmf_unif = np.full_like(x, 1 / n_obs, dtype=float)

# 3. Creación de la figura
fig, ax = plt.subplots(figsize=(8, 4.5))

# Paleta de color estilo institucional / UNAM (Azul Marino Profundo)
primary_color = "#0F4C81"
edge_color = "#082947"

# Graficado de la función de masa
bars = ax.bar(
    x,
    pmf_unif,
    width=0.6,
    color=primary_color,
    edgecolor=edge_color,
    linewidth=1.2,
    alpha=0.85,
    label=f"PMF: $P(X=x) = 1/{n_obs} = {1/n_obs:.2f}$",
)

# Línea discontinua que resalta la equiprobabilidad
ax.axhline(
    y=1 / n_obs,
    color="#E63946",
    linestyle="--",
    linewidth=1.5,
    alpha=0.7,
    label="Probabilidad constante",
)

# Personalización de ejes y cuadrícula
ax.set_xticks(x)
ax.set_ylim(0, (1 / n_obs) * 1.35)  # Espacio superior para respiración gráfica
ax.set_xlabel("Valores de la variable aleatoria ($x$)", fontweight="bold")
ax.set_ylabel("Probabilidad $P(X = x)$", fontweight="bold")

# Título y Subtítulo
ax.set_title(
    "Función de Masa de Probabilidad (PMF) - Distribución Uniforme Discreta",
    pad=15,
    fontweight="bold",
)

# Formato numérico en el eje Y a porcentaje/decimal limpio
ax.yaxis.set_major_formatter("{x:.2f}")

# Estilizado de la cuadrícula y bordes
ax.grid(True, linestyle=":", alpha=0.6)
sns.despine(top=True, right=True)

# Etiqueta de valor sobre cada barra para mayor claridad visual
for bar in bars:
    height = bar.get_height()
    ax.annotate(
        f"{height:.2f}",
        xy=(bar.get_x() + bar.get_width() / 2, height),
        xytext=(0, 3),  # Desplazamiento vertical de 3pt
        textcoords="offset points",
        ha="center",
        va="bottom",
        fontsize=7.5,
        color="#333333",
    )

# Leyenda dentro del gráfico
ax.legend(loc="upper right", frameon=True, facecolor="white", framealpha=0.9)

plt.tight_layout()
plt.show()
```

<h3>&#128187; <font color="#004D40"> Simulación en Python</font>  </3>

```python
import numpy as np
import plotly.graph_objects as go

# 1. Semilla para reproducibilidad
np.random.seed(123)

# 2. Parámetros de la distribución Uniforme Discreta (Ejemplo: Dado justo de 6 caras)
a_param = 1
b_param = 6
N_outcomes = b_param - a_param + 1  # Número de resultados posibles (6)
n_simulations = 10000

# 3. Generación de números uniformes estándar U ~ Unif(0, 1)
U = np.random.random(n_simulations)

# 4. Aplicación de la Transformación Inversa: X = a + floor(N * U)
uniformes_discretas_sim = a_param + np.floor(N_outcomes * U).astype(int)

# 5. Frecuencias relativas observadas en la simulación
unique_vals, counts = np.unique(uniformes_discretas_sim, return_counts=True)
freq_relativas = counts / n_simulations

# 6. Masa de probabilidad teórica PMF: P(X = k) = 1 / N
k_vals = np.arange(a_param, b_param + 1)
pmf_vals = np.full(shape=len(k_vals), fill_value=1.0 / N_outcomes)

# 7. Construcción de la figura interactiva con Plotly
fig = go.Figure()

# Barras de la Simulación Monte Carlo
fig.add_trace(
    go.Bar(
        x=unique_vals,
        y=freq_relativas,
        name="Simulación Monte Carlo",
        marker=dict(
            color="rgba(15, 76, 129, 0.70)",  # Azul Marino UNAM con transparencia
            line=dict(color="#082947", width=1),
        ),
    )
)

# Puntos/Líneas de la Masa Teórica PMF
fig.add_trace(
    go.Scatter(
        x=k_vals,
        y=pmf_vals,
        mode="markers+lines",
        name="Masa Teórica $P(X=k)$",
        marker=dict(color="#E63946", size=8),  # Rojo Cárdeno
        line=dict(color="#E63946", width=2, dash="dash"),
    )
)

# Personalización del diseño (Layout estilizado)
fig.update_layout(
    title=dict(
        text=f"<b>Simulación por Transformación Inversa: Uniforme Discreta UnifDiscreta({a_param}, {b_param})</b>",
        font=dict(size=16, family="sans-serif"),
        x=0.05,
    ),
    xaxis=dict(
        title="<b>Variable aleatoria discreta (k)</b>",
        tickmode="linear",
        tick0=a_param,
        dtick=1,
        showgrid=True,
        gridcolor="#E2E8F0",
        zeroline=False,
    ),
    yaxis=dict(
        title="<b>Probabilidad / Frecuencia Relativa</b>",
        showgrid=True,
        gridcolor="#E2E8F0",
        zeroline=False,
    ),
    plot_bgcolor="white",
    paper_bgcolor="white",
    bargap=0.25,
    legend=dict(
        x=0.98,
        y=0.98,
        xanchor="right",
        yanchor="top",
        bgcolor="rgba(255, 255, 255, 0.9)",
        bordercolor="#CBD5E1",
        borderwidth=1,
    ),
    width=850,
    height=480,
)

fig.show()
```

### <a id="3"></a><font color="#004D40">1.2 Distribución de Bernoulli</font>

Modela un experimento con solo dos resultados posibles: 'éxito' (1) y 'fracaso' (0).

<div class="alert alert-block alert-info">
    <b>Definición:</b> Un <b> ensayo de Bernoulli</b> se define como aquel experimento aleatorio con únicamente dos posibles resultados, llamados genéricamente: éxito y fracaso. Supondremos que las probabilidades de estos resultados son $p$ y $1-p$, respectivamente. Si se define la variable aleatoria $X$ como aquella función que lleva el resultado éxito al número 1 y el resultado fracaso al número 0, entonces decimos que $X$ tiene una <b>distribución Bernoulli</b> con parámetro $p \in (0,1)$. La función de probabilidad se puede escribir de la siguiente forma.
    $$f(x)= \begin{cases}
    1-p & \text{si  } x=0 \\
    p & \text{si  } x=1 \\
    0 & \text{en otro caso} 
    \end{cases}$$
    <div style="text-align: right; margin-top: 10px; font-size: 0.9em; color: #1A365D;">
        <i>(Rincón, 2014, p. 217)</i>
    </div>
</div>

<h3>&#128202; <font color="#004D40">Visualización en Python</font>  </3>

```python
import matplotlib.pyplot as plt
import numpy as np
import seaborn as sns
from scipy.stats import bernoulli

# Configuración del estilo general para el Notebook
sns.set_theme(style="whitegrid", palette="muted")
plt.rcParams.update(
    {
        "font.family": "sans-serif",
        "font.size": 10,
        "axes.labelsize": 11,
        "axes.titlesize": 13,
        "xtick.labelsize": 10,
        "ytick.labelsize": 9,
        "figure.dpi": 150,  # Calidad para previsualización en GitHub
    }
)

# 1. Definición de parámetros
p = 0.70  # Probabilidad de éxito
q = 1 - p  # Probabilidad de fracaso

x = np.array([0, 1])
pmf_bernoulli = bernoulli.pmf(x, p)

# 2. Creación de la figura
fig, ax = plt.subplots(figsize=(6.5, 4.5))

# Colores institucionales diferenciados para Fracaso (Rojo/Gris) y Éxito (Azul UNAM)
colors = ["#E53E3E", "#0F4C81"]
edge_colors = ["#9B2C2C", "#082947"]

# Graficado de las barras de probabilidad
bars = ax.bar(
    x,
    pmf_bernoulli,
    width=0.35,
    color=colors,
    edgecolor=edge_colors,
    linewidth=1.2,
    alpha=0.85,
)

# Personalización de etiquetas del eje X
ax.set_xticks(x)
ax.set_xticklabels(["0\n(Fracaso)", "1\n(Éxito)"], fontweight="bold")
ax.set_ylim(0, 1.0)  # Límite vertical ajustado a 1 para escala de probabilidad completa
ax.set_xlabel("Resultado del ensayo ($x$)", fontweight="bold")
ax.set_ylabel("Probabilidad $P(X = x)$", fontweight="bold")

# Título de la gráfica
ax.set_title(
    f"Función de Masa de Probabilidad (PMF) - Distribución Bernoulli ($p = {p}$)",
    pad=15,
    fontweight="bold",
)

# Formato del eje Y a decimal
ax.yaxis.set_major_formatter("{x:.2f}")

# Estilizado de cuadrícula y bordes
ax.grid(True, linestyle=":", alpha=0.6)
sns.despine(top=True, right=True)

# Anotaciones numéricas exactas sobre cada barra
for bar in bars:
    height = bar.get_height()
    ax.annotate(
        f"{height:.2f}\n({height*100:.0f}%)",
        xy=(bar.get_x() + bar.get_width() / 2, height),
        xytext=(0, 6),
        textcoords="offset points",
        ha="center",
        va="bottom",
        fontsize=9,
        fontweight="bold",
        color="#2D3748",
    )

plt.tight_layout()
plt.show()
```

### <a id="4"></a><font color="#004D40">1.3 Distribución Binomial</font>

Para experimentos que consisten en la observación de experimentos idénticos e independientes, cada uno de los cuáles puede resultar en una de dos salidas.

Por ejemplo:
- Artículos saliendo de línea de producción: pueden ser '**defectuoso**' o '**no defectuoso**'.
- Disparos en tiro al blanco: pueden ser '**acierto**' o '**no acierto**'.
- Encuesta para ver quienes están a favor de un candidato: '**a favor**' o '**en contra**'.

#### PROPIEDADES DE UN EXPERIMENTO BINOMIAL
1. Consiste en un número fijo $n$ de pruebas idénticas.
2. Cada prueba resulta en uno de dos resultados: éxito S, o fracaso F.
3. La probabilidad de éxito en una sola prueba es igual a algún valor $p$ y es el mismo de una prueba a la otra. La probabilidad de fracaso es igual a $q=(1-p)$.
4. Las pruebas son independientes.
5. La variable aleatoria de interés es $Y$, el número de éxitos observados durante las $n$ pruebas.

<h3>&#128202; <font color="#004D40">Visualización en Python</font>  </3>

```python
import matplotlib.pyplot as plt
import numpy as np
import seaborn as sns
from scipy.stats import binom

# Configuración del estilo general para el Notebook
sns.set_theme(style="whitegrid", palette="muted")
plt.rcParams.update(
    {
        "font.family": "sans-serif",
        "font.size": 10,
        "axes.labelsize": 11,
        "axes.titlesize": 13,
        "xtick.labelsize": 9,
        "ytick.labelsize": 9,
        "figure.dpi": 150,  # Alta calidad para previsualización en GitHub
    }
)

# 1. Definición de parámetros
n = 20# Número de ensayos de Bernoulli
p = 0.60  # Probabilidad de éxito
k = np.arange(0, n + 1)  # Posibles números de éxitos: {0, 1, ..., 10}

pmf_binom = binom.pmf(k, n, p)
esperanza = n * p

# 2. Creación de la figura
fig, ax = plt.subplots(figsize=(8, 4.5))

# Paleta de colores institucional UNAM
primary_color = "#0F4C81"
edge_color = "#082947"

# Graficado de las barras de probabilidad
# Se reemplazó \text por \mathrm en la leyenda para evitar el ParseFatalException
bars = ax.bar(
    k,
    pmf_binom,
    width=0.6,
    color=primary_color,
    edgecolor=edge_color,
    linewidth=1.2,
    alpha=0.85,
    label=f"$X \\sim \\mathrm{{Binom}}(n={n}, p={p})$",
)

# Línea discontinua que marca la Esperanza Matemática E[X]
ax.axvline(
    x=esperanza,
    color="#E63946",
    linestyle="--",
    linewidth=1.8,
    label=f"Esperanza $E[X] = n \\cdot p = {esperanza:.1f}$",
)

# Personalización de ejes
ax.set_xticks(k)
ax.set_ylim(0, max(pmf_binom) * 1.25)  # Margen superior para anotaciones
ax.set_xlabel("Número de éxitos ($k$)", fontweight="bold")
ax.set_ylabel("Probabilidad $P(X = k)$", fontweight="bold")

# Título de la gráfica
ax.set_title(
    f"Función de Masa de Probabilidad (PMF) - Distribución Binomial ($n={n}, p={p}$)",
    pad=15,
    fontweight="bold",
)

# Formato numérico en el eje Y
ax.yaxis.set_major_formatter("{x:.2f}")

# Estilizado de cuadrícula y bordes
ax.grid(True, linestyle=":", alpha=0.6)
sns.despine(top=True, right=True)

# Anotación numérica sobre cada barra
for bar in bars:
    height = bar.get_height()
    if height >= 0.001:  # Omitir valores insignificantes para no saturar
        ax.annotate(
            f"{height:.3f}",
            xy=(bar.get_x() + bar.get_width() / 2, height),
            xytext=(0, 4),
            textcoords="offset points",
            ha="center",
            va="bottom",
            fontsize=8,
            color="#2D3748",
        )

# Leyenda dentro del gráfico
ax.legend(loc="upper left", frameon=True, facecolor="white", framealpha=0.9)

plt.tight_layout()
plt.show()
```

### <a id="12"></a><font color="#004D40">2.3 Distribución Exponencial</font>

<h3>&#128202; <font color="#004D40">Visualización en Python</font>  </3>

```python
import matplotlib.pyplot as plt
import numpy as np
import seaborn as sns
from scipy.stats import expon

# Configuración del estilo general para GitHub Notebooks (UNAM Style)
sns.set_theme(style="whitegrid", palette="muted")
plt.rcParams.update(
    {
        "font.family": "sans-serif",
        "font.size": 10,
        "axes.labelsize": 11,
        "axes.titlesize": 13,
        "xtick.labelsize": 9,
        "ytick.labelsize": 9,
        "figure.dpi": 150,  # Alta resolución para visualización web
    }
)

# 1. Parámetros de la distribución continua
lam_rate = 0.5      # Parámetro de tasa (lambda)
scale_param = 1.0 / lam_rate # Scipy usa el parámetro de escala (1/lambda)

# 2. Rejilla de puntos x
# Definimos el rango desde 0 hasta cubrir aprox. el 99% de la probabilidad
x_max = expon.ppf(0.99, scale=scale_param)
x = np.linspace(0, x_max, 1000)

# 3. PDF y Estadísticos Teóricos
# En scipy.stats.expon: loc=0 (por defecto, inicio en 0), scale=1/lambda
pdf_expon = expon.pdf(x, scale=scale_param)
mean_theo, var_theo = expon.stats(scale=scale_param, moments="mv")
esperanza = mean_theo
desv_est = np.sqrt(var_theo)

# 4. Creación de la figura
fig, ax = plt.subplots(figsize=(8.5, 4.8))

primary_color = "#0F4C81" # Azul Marino UNAM
edge_color = "#082947"

# Graficado de la curva de densidad (PDF)
ax.plot(
    x,
    pdf_expon,
    color=primary_color,
    linewidth=2.5,
    label=rf"$X \sim \mathrm{{Exp}}(\lambda={lam_rate:.1f})$",
)

# Sombreado bajo la curva f(x)
ax.fill_between(
    x,
    pdf_expon,
    color=primary_color,
    alpha=0.20,
)

# Línea vertical para la Esperanza Matemática E[X] = 1/lambda
ax.axvline(
    x=esperanza,
    color="#E63946", # Rojo Cárdeno
    linestyle="--",
    linewidth=1.8,
    label=f"Esperanza $E[X] = 1/\lambda = {esperanza:.1f}$",
)

# Personalización de ejes
ax.set_ylim(0, max(pdf_expon) * 1.15) # Pequeño margen superior
ax.set_xlim(0, x_max)
ax.set_xlabel("Variable aleatoria continua: Tiempo ($x$)", fontweight="bold")
ax.set_ylabel("Densidad de probabilidad $f(x)$", fontweight="bold")

# Título de la gráfica
ax.set_title(
    f"Función de Densidad de Probabilidad (PDF) - Distribución Exponencial ($\lambda={lam_rate:.1f}$)",
    pad=15,
    fontweight="bold",
)

ax.yaxis.set_major_formatter("{x:.3f}")
ax.grid(True, linestyle=":", alpha=0.6)
sns.despine(top=True, right=True)

# Leyenda dentro del gráfico
ax.legend(
    loc="upper right",
    frameon=True,
    facecolor="white",
    framealpha=0.95,
    fontsize=9,
)

plt.tight_layout()
plt.show()
```

<h3>&#128187; <font color="#004D40"> Simulación en Python</font>  </3>

```python
import numpy as np
import plotly.graph_objects as go
from scipy.stats import expon

# 1. Semilla para reproducibilidad
np.random.seed(123)

# 2. Parámetros de la distribución exponencial
lam_param = 0.5  # Parámetro de tasa (lambda)
scale_param = 1.0 / lam_param  # Parámetro de escala (1/lambda)
n_simulations = 10000

# 3. Generación de números uniformes estándar U ~ Unif(0, 1)
U = np.random.random(n_simulations)

# 4. Aplicación de la Transformación Inversa: X = -ln(1 - U) / lambda
exponenciales_simuladas = -np.log(1.0 - U) / lam_param

# 5. Densidad teórica f(x) = lambda * exp(-lambda * x)
x_max = expon.ppf(0.995, scale=scale_param)
x_vals = np.linspace(0, x_max, 500)
y_vals = expon.pdf(x_vals, scale=scale_param)

# 6. Construcción de la figura interactiva con Plotly
fig = go.Figure()

# Histograma empírico normalizado como densidad de probabilidad
fig.add_trace(
    go.Histogram(
        x=exponenciales_simuladas,
        nbinsx=50,
        histnorm="probability density",
        name="Simulación Monte Carlo",
        marker=dict(
            color="rgba(15, 76, 129, 0.70)",  # Azul Marino UNAM con transparencia
            line=dict(color="#082947", width=1),
        ),
    )
)

# Curva de la densidad teórica f(x)
fig.add_trace(
    go.Scatter(
        x=x_vals,
        y=y_vals,
        mode="lines",
        name="Densidad Teórica $f(x)$",
        line=dict(color="#E63946", width=3),  # Rojo Cárdeno
    )
)

# Personalización del diseño (Layout estilizado)
fig.update_layout(
    title=dict(
        text=f"<b>Simulación por Transformación Inversa: Exponencial Exp(&lambda;={lam_param:.1f})</b>",
        font=dict(size=16, family="sans-serif"),
        x=0.05,
    ),
    xaxis=dict(
        title="<b>Variable aleatoria (x)</b>",
        showgrid=True,
        gridcolor="#E2E8F0",
        zeroline=False,
    ),
    yaxis=dict(
        title="<b>Densidad de Probabilidad</b>",
        showgrid=True,
        gridcolor="#E2E8F0",
        zeroline=False,
    ),
    plot_bgcolor="white",
    paper_bgcolor="white",
    bargap=0.05,
    legend=dict(
        x=0.98,
        y=0.98,
        xanchor="right",
        yanchor="top",
        bgcolor="rgba(255, 255, 255, 0.9)",
        bordercolor="#CBD5E1",
        borderwidth=1,
    ),
    width=850,
    height=480,
)

fig.show()
```

### <a id="13"></a><font color="#004D40">2.4 Distribución Gamma</font>

<h3>&#128202; <font color="#004D40">Visualización en Python</font>  </3>

```python
import matplotlib.pyplot as plt
import numpy as np
import seaborn as sns
from scipy.stats import gamma

# Configuración del estilo general
sns.set_theme(style="whitegrid", palette="muted")
plt.rcParams.update(
    {
        "font.family": "sans-serif",
        "font.size": 10,
        "axes.labelsize": 11,
        "axes.titlesize": 13,
        "xtick.labelsize": 9,
        "ytick.labelsize": 9,
        "figure.dpi": 150,
    }
)

# 1. Parámetros de la distribución continua
alpha_shape = 3.0  # Parámetro de forma (a)
theta_scale = 2.0  # Parámetro de escala (scale)

# 2. Rejilla de puntos x (cubre > 99.5% de la probabilidad)
x_max = gamma.ppf(0.995, a=alpha_shape, scale=theta_scale)
x = np.linspace(0, x_max, 1000)

# 3. PDF y Estadísticos Teóricos
# En scipy.stats.gamma: a = alpha, scale = theta
pdf_gamma = gamma.pdf(x, a=alpha_shape, scale=theta_scale)
mean_theo, var_theo = gamma.stats(a=alpha_shape, scale=theta_scale, moments="mv")
esperanza = mean_theo
varianza = var_theo
desv_est = np.sqrt(varianza)

# 4. Creación de la figura
fig, ax = plt.subplots(figsize=(8.5, 4.8))

primary_color = "#0F4C81"
edge_color = "#082947"

# Graficado de la curva de densidad (PDF)
ax.plot(
    x,
    pdf_gamma,
    color=primary_color,
    linewidth=2.5,
    label=rf"$X \sim \mathrm{{Gamma}}(\alpha={alpha_shape:.1f}, \theta={theta_scale:.1f})$",
)

# Sombreado bajo la curva f(x)
ax.fill_between(
    x,
    pdf_gamma,
    color=primary_color,
    alpha=0.20,
)

# Línea vertical para la Esperanza Matemática E[X]
ax.axvline(
    x=esperanza,
    color="#E63946",
    linestyle="--",
    linewidth=1.8,
    label=f"Esperanza $E[X] = \\alpha\\theta = {esperanza:.1f}$",
)

# Personalización de ejes
ax.set_ylim(0, max(pdf_gamma) * 1.2)
ax.set_xlim(0, x_max)
ax.set_xlabel("Variable aleatoria continua ($x$)", fontweight="bold")
ax.set_ylabel("Densidad de probabilidad $f(x)$", fontweight="bold")

# Título de la gráfica
ax.set_title(
    f"Función de Densidad de Probabilidad (PDF) - Distribución Gamma ($\\alpha={alpha_shape:.1f}, \\theta={theta_scale:.1f}$)",
    pad=15,
    fontweight="bold",
)

ax.yaxis.set_major_formatter("{x:.3f}")
ax.grid(True, linestyle=":", alpha=0.6)
sns.despine(top=True, right=True)

# Leyenda dentro del gráfico
ax.legend(
    loc="upper right",
    frameon=True,
    facecolor="white",
    framealpha=0.95,
    fontsize=9,
)

plt.tight_layout()
plt.show()
```

## <a id="16"></a><font color="#2F855A">5. Ley de los grandes Números</font>

Desde el siglo XVI, Gerolano Cardano observó que la precisión de las estadísticas empíricas incrementa con el número de ibservaciones. Bernoulli y Poisson retomaron esa idea y la retomaron con la ley de los grandes números, misma que posteriormente fue generalizada por matemáticos dedicados a la probabilidad (Rincón, 2014).

<div style="background-color: #F3E8FF; padding: 15px; border-radius: 6px; margin: 15px 0;">
    <b style="color: #581C87; font-size: 1.05em;">Teorema (Ley de los grandes números):</b> 
    Sea $X_1, X_2, \dots$ una sucesión infinita de variables aleatorias independientes e idénticamente distribuidas con media finita $\mu$. Entonces, cuando $n \to \infty$,
    $$\frac{1}{n} \sum_{i=1}^{n} X_i \to \mu$$
    en donde la convergencia se verifica en el sentido casi seguro (ley fuerte) y también en probabilidad (ley débil).
    <div style="text-align: right; margin-top: 10px; font-size: 0.9em; color: #6B21A8;">
        <i>(Rincón, 2014, p. 368)</i>
    </div>
</div>

## <a id="17"></a><font color="#2F855A">6. Teorema Central del Límite</font>

<div style="background-color: #F3E8FF; padding: 15px; border-radius: 6px; margin: 15px 0;">
    <b style="color: #581C87; font-size: 1.05em;">Teorema (Teorema central del límite):</b> 
    Sea $X_1, X_2, \dots$ una sucesión infinita de variables aleatorias independientes e idénticamente distribuidas con media $\mu$ y varianza finita $\sigma^2$. Entonces la función de distribución de la variable aleatoria 
    $$ Z_n = \frac{(X_1+\dots+X_n)-n\mu}{\sqrt{n\sigma^2}} $$
    tiende a la función de distribución normal estándar cuando $n$ tiende a infinito.
    <div style="text-align: right; margin-top: 10px; font-size: 0.9em; color: #6B21A8;">
        <i>(Rincón, 2014, p. 376)</i>
    </div>
</div>
