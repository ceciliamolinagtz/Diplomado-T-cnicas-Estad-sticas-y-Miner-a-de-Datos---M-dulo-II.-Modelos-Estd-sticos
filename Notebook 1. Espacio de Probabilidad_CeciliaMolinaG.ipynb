## <a id="1"></a><font color="#2F855A">1. Definición de espacio de probabilidad</font>

<div class="alert alert-block alert-info">

## <a id="2"></a><font color="#2F855A">2. Espacio muestral $\Omega$</font> 

<div class="alert alert-block alert-info">
<b>Definición:</b> El <b>espacio muestral</b>, también llamado espacio muestra, de un experimento aleatorio es el conjunto de todos los posibles resultados del experimento y se le denota, generalmente, por la letra griega $\Omega$. A un resultado particular del experimento se le denota por la letra $\omega$. 
    <div style="text-align: right; margin-top: 10px; font-size: 0.9em; color: #1A365D;">
        <i>(Rincón, 2014, p. 5)</i>
    </div>
</div>

<p>&#128161; Todos los resultados del experimento deben estar dentro del espacio muestral.</p> 

<h3>&#127922; <font color="#004D40">Ejemplo</font>  </3>

Al lanzar un dado, el `espacio muestral` está dado por cada resultado posible, es decir, cada una de las caras del dado. 

Por lo tanto $$\Omega = \{ 1,2,3,4,5,6\}$$

Los resultados particulares están dados por cada una de las caras:

- Cae 1 $\to$ $\omega_1 = {1}$
- Cae 2 $\to$ $\omega_2 = {2}$
- Cae 3 $\to$ $\omega_3 = {3}$
- Cae 4 $\to$ $\omega_4 = {4}$
- Cae 5 $\to$ $\omega_5 = {5}$
- Cae 6 $\to$ $\omega_6 = {6}$

Como podemos observar, todos los resultados posibles están dentro de este espacio muestral.

Dicho con otras palabras, los `eventos` son los resultados que se dan al hacer un experimento (Wackerly et al.,2010).

## <a id="4"></a><font color="#2F855A">4. Axiomas de Kolmogorov</font> 

<div class="alert alert-block alert-info">
<b>AXIOMAS DE LA PROBABILIDAD</b>
\begin{equation}
    1. \text{ } P(A) \geq  0.
\end{equation}

\begin{equation}
    2. \text{ } P(\Omega) =  1.
\end{equation}

\begin{equation}
    3. \text{ }  P(\bigcup_{i=1}^{\infty}{A_k}) = \sum_{i=1}^{\infty}{P(A_k)} \text{ cuando } A_1, A_2, \dots \text{son ajenos dos a dos}.
\end{equation}
    <div style="text-align: right; margin-top: 10px; font-size: 0.9em; color: #1A365D;">
        <i>(Rincón, 2014, p. 36)</i>
    </div>
</div>

Expresado con palabras, el primer axioma de Kolmogorov nos dice que las probabilidades negativas no son posibles, la frecuencia relativa de que ocurra cualquier evento siempre debe ser mayor o igual a cero. En segundo lugar, la suma de las probabilidades de todos los eventos dados en un espacio muestral debe ser uno, y finalmente, si dos eventos son mutuamente excluyentes, la frecuencia relativa de la unión de ambos es la suma de sus respectivas frecuencias relativas (Wackerly et al., 2010).

<p>&#128161; Observemos que la suma de todos los eventos da un 1 porque todos los resultados posibles dan el 100% de ocurrencias. </p> 

<h3>&#127922; <font color="#004D40">Ejemplo</font>  </3>

Siguiendo con el ejemplo del dado, imaginemos que tenemos un dado justo. Es decir, que cada cara tiene exactamente la misma probabilidad de salir. 

De la primera parte del ejemplo tenemos que $\Omega = \{1,2,3,4,5,6\}$, es decir, existen $6$ posibles resultados. Como es un dado justo, cada uno de los resultados tiene la misma probabilidad de salir, y por los axiomas de Kolmogorov sabemos que $P(\Omega)= 1$.

Considerando que $|\Omega|=6$.

Entonces, cada resultado tiene una probabilidad de $\frac{1}{6}$ de salir.

Por lo tanto, tenemos que:
- $P(\omega_1) = P(\omega_2) = \dots = P(\omega_6) = \frac{1}{6}$.

Esta es una manera de ver en la práctica con un ejemplo sencillo los axiomas. Todos los resultados individuales tienen 0 o más probabilidades de ocurrir (específicamente $\frac{1}{6}$ de probabilidad). La suma de las probabilidades de todos los resultados da 1, y la probabilidad de suceder de la unión de 2 resultados ajenos, es la suma de ambas probabilidades.

## <a id="5"></a><font color="#2F855A">5. Probabilidad condicional</font> 

A veces, la probabilidad de que ocurra un evento depende de si han ocurrido otros eventos o no: "la probabilidad condicional de un evento es la probabilidad (frecuencia relativa de ocurrencia) del evento dado el hecho de que uno o más eventos ya han ocurrido" (Wackerly et al, 2010, p. 52).

<div class="alert alert-block alert-info">
<b>Definición:</b> Sean $A$ y $B$ dos eventos y supongamos que $B$ tiene probabilidad estrictamente positiva. La <b>probabilidad condicional</b> del evento $A$, dado el evento $B$, se denota por el símbolo $P(A|B)$ y se define como el cociente
    
\begin{equation}
    P(A|B) = \frac{P(A \cap B)}{P(B)}
\end{equation}
    <div style="text-align: right; margin-top: 10px; font-size: 0.9em; color: #1A365D;">
        <i>(Rincón, 2014, p. 73)</i>
    </div>
</div>

<h3>&#127922; <font color="#004D40">Ejemplo</font>  </3>

Veamos cómo se puede ver la probabilidad condicional tomando como ejemplo el dado.

Supongamos que se tira un dado justo una vez. Sabemos que se obtuvo un número impar. 

¿Cuál es la probabilidad de que el resultado sea 1?

Tenemos dos eventos definidos aquí. 
Sean:
- $A$: observar un 1
- $B$: observar un número impar

Buscamos la probabilidad del evento $A$ dado que $B$ ha ocurrido. El evento $A \cap B$ requiere que se observen un 1 y un número impar. En este caso, el 1 está contenido dentro de 'número impar', por lo que $A \subset B$, es decir, $A \cap B = A$.

Entonces tenemos por un lado que: 

$P(A \cap B) = P(A) = P(\omega_1) = \frac{1}{6}$

y por el otro:

$P(B) = P(\omega_1)+P(\omega_3)+P(\omega_5) = \frac{1}{6} + \frac{1}{6} + \frac{1}{6} = \frac{3}{6} = \frac{1}{2}$

Utilizando la definición de evento condicional, tenemos que:

\begin{equation}
    P(A|B) = \frac{P(A \cap B)}{P(B)} = \frac{\frac{1}{6}}{\frac{1}{2}} = \frac{1}{3}
\end{equation}

## <a id="6"></a><font color="#2F855A">6. Independencia</font>

Por otro lado, tenemos el caso donde la probabilidad de que ocurra un evento $A$ no sea afectada por el hecho de que ocurra o no ocurra un evento $B$. 

<div class="alert alert-block alert-info">
<b>Definición:</b> Se dice que los eventos $A$ y $B$ son independientes si se cumple la igualdad
    
\begin{equation}
    P(A \cap B) = P(A)P(B).
\end{equation}
    <div style="text-align: right; margin-top: 10px; font-size: 0.9em; color: #1A365D;">
        <i>(Rincón, 2014, p. 92)</i>
    </div>
</div>

De acuerdo con Wackerly et al (2010), para que dos eventos sean independientes también se pueden considerar los casos: 
- $P(A|B) = P(A)$
- $P(B|A) = P(B)$

<p>&#128161; Es importante no confundir la independencia son que dos eventos sean mutuamente excluyentes. De hecho, dos eventos con probabilidad mayor a cero que son <b>mutuamente excluyentes</b>, es decir, que no pueden ocurrir al mismo tiempo, <b>SIEMPRE</b> son dependientes. </p> 

<h3>&#127922; <font color="#004D40">Ejemplo</font>  </3>

Para ejemplificar esto observemos los siguientes eventos:
- $A$: observar un número impar
- $B$: observar un número par
- $C$: observar un 1 o 2

Tenemos que:

$P(A)= \frac{1}{2}$

$P(B)= \frac{1}{2}$

$P(C)= \frac{1}{3}$

<b>a) ¿A y B son independientes?</b>

Observemos que $A \cap B = \emptyset$ , $P(A|B)=0$ y $P(A|B) \neq P(A)$, al no cumplir con la definición, tenemos que $A$ y $B$ son dependientes.

<b>b) ¿A y C son independientes?</b>

Observemos que $P(A|C) = \frac{1}{2}$ y, $P(A)= \frac{1}{2}$, por lo tanto $P(A|C) =P(A)$ y podemos concluir que los eventos son independientes.

## <a id="7"></a><font color="#2F855A">7. Enfoque clásico</font> 

<div class="alert alert-block alert-info">
<b>Definición:</b> Sea $A$ un subconjunto de un espacio muestral $\Omega$ de cardinalidad finita. Se define la <b>probabilidad clásica</b> del evento $A$ como el cociente
    $$ P(A) = \frac{|A|}{|\Omega|} $$
    <div style="text-align: right; margin-top: 10px; font-size: 0.9em; color: #1A365D;">
        <i>(Rincón, 2014, p. 20)</i>
    </div>
</div>

Es importante resaltar que para poder utilizar esta definición. necesitamos que el conjunto $\Omega$ sea `finito` y `equiprobable`; es decir, que todos los elementos de $A$ tengan la misma probabilidad de ocurrir (tengan el mismo peso).

## <a id="8"></a><font color="#2F855A">8. Enfoque frecuentista</font> 

<div class="alert alert-block alert-info">
<b>Definición:</b> Sea $n_A$ el número de ocurrencias de un evento $A$ en $n$ realizaciones de un experimento aleatorio. La probabilidad frecuentista del evento $A$ se define como el límite
    $$ P(A) = \lim_{n \to \infty}\frac{n_a}{n} .$$
    <div style="text-align: right; margin-top: 10px; font-size: 0.9em; color: #1A365D;">
        <i>(Rincón, 2014, p. 32)</i>
    </div>
</div>

No es posible hacer un número infinito de veces un experimento, pero esta definición nos permite tener una aproximación empírica de $P(A)$.

## <a id="9"></a><font color="#2F855A">9. Enfoque geométrico</font> 

<div class="alert alert-block alert-info">
<b>Definición:</b> Si un experimento aleatorio tiene como espacio muestral $\Omega \subset \mathbb{R}^2$ cuya área está bien definida y es finita, entonces se define la probabilidad geométrica de un evento $A \subseteq \Omega$ como
    $$ P(A) = \frac{\text{Área de } A}{\text{Área de } \Omega} ,$$
cuando el concepto de área del subconjunto $A$ está bien definido.
    <div style="text-align: right; margin-top: 10px; font-size: 0.9em; color: #1A365D;">
        <i>(Rincón, 2014, p. 23)</i>
    </div>
</div>

Para poder aplicar la fórmula anterior, es necesario que la probabilidad de observar la ocurrencia de un evento $A$ dependa únicamente de su área y no del conjunto mismo.

## <a id="10"></a><font color="#2F855A">10. Problema del encuentro</font> 

Dos personas planean encontrarse en un lugar específico, en un rango de una hora $[0,60]$ minutos (1 hora).
- Si A llega primero, entonces piensa esperar 10 minutos.
- Si B llega primero, entonces piensa esperar 10 minutos.

Se elige un número al azar entre $[0,60]$ y supongamos que el par de números se elige sobre el cuadrado de modo que la probabilidad de un evento sea el área favorable entre el área total.

Sea 

- $A$: las dos personas se encuentran a una distancia de 10 minutos.

Nos gustaría encontrar la probabilidad de $A$.

```python
import numpy as np
import matplotlib.pyplot as plt # para graficar

# Vamos a generar 1000 simulaciones
muestra = 1000

# Generamos los tiempos aleatorios entre [0,60]
x = np.random.uniform(0,60, muestra)
y = np.random.uniform(0,60, muestra)

# Añadimos una condición de encuentro
encuentro = np.where(abs(x-y) <= 10, 'r', 'b') # rojo si se encuentra, azul si no.

# Ahora calculamos la probabilidad de que se encuentren
k=0
for elemento in encuentro:
    if elemento == 'r': #los favorables
        k += 1 # lo añada
print('La probabilidad del evento A es: {}'.format(k/muestra))

# Graficamos
plt.figure(figsize=(5,5)) # tamaño de la figura
plt.scatter(x,y, c=encuentro, s=1)
plt.show()
```

## <a id="11"></a><font color="#2F855A">11. Ejemplo del volado</font> 

Sea el experimento aleatorio de lanzar una moneda (no justa).

```python
import random # para generar números aleatorios
def volado(p):
    return 'sol' if random.random() < p else 'águila'

#random.random() genera un número entre 0 y 1
```

```python
volado(0.4) # hay un 40% de obtener sol
```

```python
# Vamos a realizar 10 volados con probabilidad p
# range(n) crea un contador que va del 0 al 9 (repite n veces)
vol = [volado(0.4) for i in range(10)]
vol
```

```python
# Para contar el número de soles en una lista
vol.count('sol')
```

```python
#Para contar el número de águilas
vol.count('águila')
```

Vamos a simular la probabilidad frecuentista, es decir, vamos a realizar un experimento con 1000 volados (individual) y puedo repetirlo 500 veces (500 personas).

```python
# Creamos una lista para almacenar las probabilidades
p = []

for i in range(500): # realizo 500 experimentos
    N = 1000 # cada experimento tiene 1000 volados
    resultados = [] # se almacenan 'sol' o 'águila'
    
    for j in range(N):
        resultado = volado(0.63)
        resultados.append(resultado)
        
    n_soles = resultados.count('sol')
    p_soles = n_soles / N # probabilidad frecuentista
    
    # Añadimos estas probabilidades a la lista p
    p.append(p_soles)
    
    # Probabilidad frecuentista promedio
    p_soles_prom = sum(p) / 500
    
p_soles_prom
```

## <a id="12"></a><font color="#2F855A">12. Ejemplo adicional: El problema del palo roto</font> 

Imagina que tienes una barra o palo de longitud $1$ y la cortas en dos puntos al azar, $x$ e $y$ (ambos entre 0 y 1). Esto la divide en 3 pedazos.

¿Cuál es la probabilidad de que con esos 3 pedazos se pueda formar un triángulo?

**Condición geométrica:**
Para que tres segmentos formen un triángulo, la desigualdad triangular exige que **ningún pedazo mida más de la mitad de la barra (0.5)**.

Si llamamos al primer corte $x$ y al segundo corte $y$ (asumiendo $x < y$), las longitudes son:
1. Primer pedazo: $x$
2. Segundo pedazo: $y - x$
3. Tercer pedazo: $1 - y$

La probabilidad teórica exacta de que se pueda formar un triángulo es del **25%** ($0.25$).

```python
import numpy as np

# 1. Número de experimentos
N = 10000

# 2. Generar 2 puntos de corte al azar entre 0 y 1 para cada experimento
puntos = np.random.uniform(0, 1, (N, 2))

# Ordenar los cortes para que x sea el menor e y el mayor
x = np.min(puntos, axis=1)
y = np.max(puntos, axis=1)

# 3. Calcular la longitud de los 3 pedazos
pedazo1 = x
pedazo2 = y - x
pedazo3 = 1 - y

# 4. Condición: Ningún pedazo puede medir 0.5 o más
forma_triangulo = (pedazo1 < 0.5) & (pedazo2 < 0.5) & (pedazo3 < 0.5)

# 5. Resultados
prob_simulada = np.mean(forma_triangulo)

print(f"Experimentos realizados: {N:,}")
print(f"Casos donde SE forma un triángulo: {np.sum(forma_triangulo):,}")
print(f"Probabilidad simulada: {prob_simulada:.4f}")
print(f"Probabilidad teórica: 0.2500")
```

```python
import numpy as np
import matplotlib.pyplot as plt # para graficar

# Vamos a generar 1000 simulaciones
muestra = 1000

# Generamos dos puntos de corte aleatorios entre [0,1]
x_raw = np.random.uniform(0, 1, muestra)
y_raw = np.random.uniform(0, 1, muestra)

# Ordenamos x e y para que x sea siempre el primer corte e y el segundo
x = np.minimum(x_raw, y_raw)
y = np.maximum(x_raw, y_raw)

# Calculamos la longitud de los 3 pedazos
p1 = x
p2 = y - x
p3 = 1 - y

# Añadimos una condición para formar el triángulo (ningún pedazo >= 0.5)
triangulo = np.where((p1 < 0.5) & (p2 < 0.5) & (p3 < 0.5), 'r', 'b') # rojo si forma triángulo, azul si no.

# Ahora calculamos la probabilidad de que se forme el triángulo
k=0
for elemento in triangulo:
    if elemento == 'r': #los favorables
        k += 1 # lo añada
print('La probabilidad del evento A es: {}'.format(k/muestra))

# Graficamos
plt.figure(figsize=(5,5)) # tamaño de la figura
plt.scatter(x,y, c=triangulo, s=1)
plt.show()
```
